# Principal Component Analysis

When you have a very wide dataset, you can reduce its dimensionality using principal component analysis (PCA). PCA relies on the concepts of linear transformations and singular value decomposition (SVD) in linear algebra.  
Luckily, scikit-learn provides a package that we can use to perform PCA instead of worrying about the linear algebra. You can find the documentation [here](https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.PCA.html).

## Part 1: Setup
This first part will serve as the setup for the full toolkit. Think ,for example, of which file you want to use for data visualization.

```python
# Load packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from pathlib import Path
```

## Variables selection

```python
# Changes the number of principal components used in PCA. (2 or 3)
N_COMPONENTS = 3
```

## Reads file
Please have only one csv file in the folder.

```python
csv_file = sorted(Path.cwd().glob("*.csv"))
if len(csv_file) != 1:
    raise FileNotFoundError(
        f"Expected one CSV file in {Path.cwd()}, found: {[f.name for f in csv_file
    ]}"
    )
csv_file = csv_file[0]

print(f"Reading: {csv_file}")
```

Loads the data from the dataset

```python
# Load data from the csv file
df = pd.read_csv(csv_file, index_col=0)
```

Shows a short description of the dataset.

```python
df.describe()
```

```python
# TODO: Let user choose which 2 groups to select for further PCA analysis
```

```python
# Use all measurement columns, including the first measurement stored in the index.
X = np.column_stack((df.index.to_numpy(), df.iloc[:, :-1].to_numpy()))

# The last column contains the target labels.
y = df.iloc[:, -1].to_numpy()
```

```python
# Reduce the selected numeric measurements to two coordinates per row.
pca = PCA(n_components=N_COMPONENTS)
principal_comps = pca.fit_transform(X)

principal_df = pd.DataFrame(principal_comps, columns=[f'pc{i+1}' for i in range(N_COMPONENTS)])

# Keep labels only when the original dataset provides them.
if y is not None:
    principal_df['target'] = y
```

```python
principal_df.head()
```

```python
plt.style.use("ggplot")

if N_COMPONENTS == 3:
    fig = plt.figure(figsize=(12, 5))

    ax2d = fig.add_subplot(1, 2, 1)
    ax2d.scatter(principal_df["pc1"], principal_df["pc2"])
    ax2d.set_xlabel("Principal component 1")
    ax2d.set_ylabel("Principal component 2")
    ax2d.set_title("PCA — 2D")

    ax3d = fig.add_subplot(1, 2, 2, projection="3d")
    ax3d.scatter(
        principal_df["pc1"],
        principal_df["pc2"],
        principal_df["pc3"].to_numpy(),
    )  # pyright: ignore[reportArgumentType]
    ax3d.set_xlabel("Principal component 1")
    ax3d.set_ylabel("Principal component 2")
    ax3d.set_zlabel("Principal component 3")
    ax3d.set_title("PCA — 3D")

elif N_COMPONENTS == 2:
    fig, ax2d = plt.subplots(figsize=(6, 5))
    ax2d.scatter(principal_df["pc1"], principal_df["pc2"])
    ax2d.set_xlabel("Principal component 1")
    ax2d.set_ylabel("Principal component 2")
    ax2d.set_title("PCA — 2D")

else:
    raise ValueError("N_COMPONENTS must be 2 or 3.")

plt.tight_layout()
plt.show()
```
